```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Dataset loading
data = pd.read_csv('Superstore.csv', encoding='latin-1')
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data = data.dropna(subset = ['Postal Code'])
```

```python
data.info()
```

```python
#Применяем функцию к столбцам с датами
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date']  = pd.to_datetime(data['Ship Date'])

#Проверяем, не появилось ли пропусков в данных после преобразования
print('Пропусков в Order Date:', data['Order Date'].isna().sum()) 
print('Мин/макс:', data['Order Date'].min(), data['Order Date'].max())
```

```python
#Разделим столбец Order Date на Order Day, Order Month и Order Year для дальнейшего удобства анализа
data['Order Year']  = data['Order Date'].dt.year
data['Order Month'] = data['Order Date'].dt.month
data['Order Day']   = data['Order Date'].dt.day
```

```python
#Так же добавим столбцы со значениями дня недели и квартала
data['Order Quarter'] = data['Order Date'].dt.quarter
data['Order DayOfWeek'] = data['Order Date'].dt.dayofweek
data['Order YearMonth'] = data['Order Date'].dt.to_period('M')
```

```python
print(data[['Order Date', 'Order Year', 'Order Month', 'Order Day',
          'Order DayOfWeek', 'Order Quarter','Order YearMonth']].head())
print(data.dtypes)
```

```python
#Находим и выводим в новый столбец время, в днях, затраченное на доставку
data['Shipping days'] = (data['Ship Date'] - data['Order Date'])
```

```python
print(data['Shipping days'].head())
```

```python
from datetime import datetime, timedelta
#Проверим, нет ли отрицательных строк в датасете
neg = data[data['Shipping days'] < timedelta(0)][['Order Date', 'Ship Date', 'Shipping days']]
print(neg.head(20))
print('Всего отрицательных:', len(neg))
print('Доля от всех строк: {:.2%}'.format(len(neg) / len(data)))
```

```python
#Смотрим на распределение полученных результатов чтобы проверить их подлинность 
print(data['Shipping days'].describe())
print(data['Shipping days'].value_counts().sort_index().head(10))
```

```python
#Проверям орицательные значения в столбцах Sales и Quantity(обычно это возвраты)
print(f"Количество отрицательных значений в столбце продаж: {(data['Sales']<0).sum()}")
print(f"Количество отрицательных значений в столбце количетсва проданных товаров: {(data['Quantity']<0).sum()}")
```

```python
#Считаем прибыльную маржу и выводим в новый столбец, значение будет в процентах
data['Profit Margin'] = (data['Profit']/data['Sales']*100).round(2)
data['Profit Margin']
```

```python
#Проверяем на дубликаты
data.duplicated().sum()
```

```python
#В итоге мы получили очищенный датафрейм, подготовленный для анализа, переведём его в csv формат и продолжим анализ в Google Sheets
data.to_csv("Clearead_store_data.csv", index = False)
```
